# Part A — Decoder-Only Model

## Decoder-only example: GPT-style Text Generation

A decoder-only model predicts the next token based on previous tokens.

We will use **DistilGPT-2**, a small GPT-style model.


````python
generator = pipeline("text-generation", model="distilgpt2")

prompt = "Natural language processing is"

output = generator(prompt, max_new_tokens=30, do_sample=False)

print(output[0]["generated_text"])
````

# Part B — Encoder–Decoder Model

## Encoder–decoder example: Instruction / Text-to-Text Generation

Encoder–decoder models contain two main parts:

1. The **encoder** reads the input.
2. The **decoder** produces the output.

We will use **FLAN-T5-small**.


````python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small")

prompt = "Translate to French: I love natural language processing."

# Convert text into tokens
output = model.generate(**tokenizer(prompt, return_tensors="pt"))

# Convert tokens back to text
print(tokenizer.decode(output[0], skip_special_tokens=True))
````

# Part C — Encoder-Only Model

## Encoder-only example: Sentiment Classification

Encoder-only models read the full input and create a representation of it.

They are very useful when the task is to **understand or classify text**.

Here we use a DistilBERT sentiment model.


````python
classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")

text = "This NLP lab is easy and interesting."

result = classifier(text)
print(result)
````

````python
from transformers import pipeline

classifier = pipeline("sentiment-analysis",
                      model="distilbert-base-uncased-finetuned-sst-2-english")

sentences = [
    "The movie was excellent.",
    "The lecture was difficult to understand.",
    "I enjoyed learning about transformers."
]

results = classifier(sentences)

for sentence, result in zip(sentences, results):
    print(f"{sentence}\n  Label: {result['label']}, Score: {result['score']:.4f}\n")
````

## Task 3 — Encoder–Decoder Model

Use the FLAN-T5 model to perform the following:

- Summarize a short paragraph.


````python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "google/flan-t5-base"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

paragraph = """Transformers are a type of deep learning model introduced in 2017.
They rely on a mechanism called self-attention, which allows the model to weigh
the importance of different words in a sentence. Unlike recurrent neural networks,
transformers process all words in parallel, which makes training much faster.
Today, transformers are the foundation of most modern NLP systems, including
machine translation, chatbots, and text summarization tools."""

prompt = "Summarize the following text: " + paragraph

inputs = tokenizer(prompt, return_tensors="pt")
outputs = model.generate(**inputs, max_new_tokens=60)
summary = tokenizer.decode(outputs[0], skip_special_tokens=True)

print("Original:\n", paragraph)
print("\nSummary:\n", summary)
````


## Task 5 — Short Reflection

Answer in 3–5 sentences:

**Why is a decoder-only model suitable for a chatbot such as ChatGPT?**


Tsak 5:

A decoder-only model is trained to predict the next word based on the previous words. This matches how a chatbot works, since it reads the conversation and generates a reply one word at a time. It uses masked self-attention, so each word only attends to the words before it. The whole conversation can be treated as one continuous sequence, so a separate encoder is not needed. Also, these models scale well with large data, allowing them to handle many tasks through prompts alone.

### Question

Why do you sometimes get different outputs even when the prompt is the same?

**Hint:** Sampling introduces randomness when choosing the next token.


````python
from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")

prompt = "Artificial intelligence will"

for i in range(3):
    output = generator(prompt, max_new_tokens=30, do_sample=True,
                       temperature=0.9, top_k=50)
    print(f"Run {i+1}: {output[0]['generated_text']}\n")
````

Because sampling randomly picks the next token based on its probability, not always the most likely one. A different choice at any step changes the rest of the text. Without `set_seed(42)`, the randomness is not fixed, so each run gives a different output.
